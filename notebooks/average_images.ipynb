```python
import SimpleITK as sitk
from pathlib import Path
from tqdm.auto import tqdm
```

```python
inp_folder = Path("/home/varhai/PRIMUS/data/83_BIOINFORMATICS/Igor/mouse_skull_workdir")
#read text of sorted images
input_images = Path("/home/varhai/PRIMUS/data/83_BIOINFORMATICS/Igor/workdir/images_4_average.txt").read_text()
input_images = input_images.split("\n")
image_files = [(inp_folder /x/"coregistered_head.nii.gz") for x in input_images ]


images = []
c_num = 21
divide = len(image_files)
res_im= None
for imag_f in tqdm(image_files):
    imag = sitk.ReadImage(imag_f)
    arr = sitk.GetArrayFromImage(imag)
    
    if res_im is None:
        res_im = arr / divide
    else:
        res_im += arr/ divide
        

```

```python
x = sitk.ReadImage(image_files[0])
res_x =  sitk.GetImageFromArray(res_im)
res_x.SetDirection(x.GetDirection())
res_x.SetOrigin(x.GetOrigin())
res_x.SetSpacing(x.GetSpacing())
#res_x.SetPixel(sitk.sitkFloat32)
res_x= sitk.Cast(res_x,sitk.sitkFloat32)
sitk.WriteImage(res_x, '/home/varhai/PRIMUS/data/83_BIOINFORMATICS/Igor/reference_image/average_image.nii.gz',useCompression=True)
```
