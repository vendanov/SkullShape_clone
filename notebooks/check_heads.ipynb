```python
from configparser import ConfigParser
from typing import Union, List, Dict
from PIL import Image
import SimpleITK as sitk
# deve
def get_subjects(fold : Union[str,pathlib.Path]) -> List[pathlib.Path]:
    """
    parse folder and get all mice folders inside
    :param fold: folder with cohort
    :return: folder with mouse
    """
    i_f = fold
    if isinstance(fold,str):
        i_f = pathlib.Path(i_f)
    res = []
    for file_ in i_f.iterdir():
        if file_.is_file():
            continue
        #filter
        if file_.name.startswith('BMD_TMD') or file_.name.startswith('Morphology') or file_.name.startswith('phantom'):
            continue
        res.append(file_)
    return res
```

```python
heads = []
for fold in tqdm(list(pathlib.Path(input_folders).iterdir())):
    if fold.is_file():
        continue
    subjs = get_subjects(fold)

    for subj in subjs:
        head_f = None
        for x in subj.glob("**/*_head"):
            head_f = x
            break
        if head_f is not None:
            headc = len(list(head_f.iterdir()))
            heads.append(  [ subj , headc ]  )
```

```python
head_sizes = pd.DataFrame(heads)
```
