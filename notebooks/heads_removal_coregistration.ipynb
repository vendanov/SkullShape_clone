Removal of the files where coregistration failed before cohort 316

```python
import os
import shutil
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
```

```python

no_cohors = ['c' + str(i) for i in range(316,1000)]
base_folder = Path(r"/home/varhai/PRIMUS/data/83_IMPC_DATA/Bioimaging/SCREENad.Morpho&BodyComp")

for fold_ in base_folder.iterdir():
    # if 
    
    if fold_.is_file():
        continue
        
    # if c316 or  higher in the name then continue
    
    if any([i in fold_.name for i in no_cohors]):
        print("Skipping cohort",fold_)
        continue
    file_ = (fold_/"abnormality_automated_combined.csv")
    
    if not file_.exists():
        continue
    #print("q")
    file_analysis = pd.read_csv(file_,index_col=0)
    tmp = file_analysis[file_analysis['proc_dist'] > 8000]
    if tmp.shape[0] > 0:
        # remove the file_
        os.remove(file_)
        os.remove(fold_/'logs_skull.txt')
        
        # remove subfold files
        for subj in tmp.index:
            shutil.rmtree(fold_/'skull_shape_workdir'/subj)
            print("Removed",fold_/'skull_shape_workdir'/subj)
    
        
        
    #print(file_analysis)
    #print(td)
```
